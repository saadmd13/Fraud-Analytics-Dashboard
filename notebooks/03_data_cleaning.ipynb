# Data Cleaning

## Objective

Prepare a clean, consistent dataset for exploratory data analysis and machine learning.

### Cleaning Tasks

- Load Dataset
- Verify Missing Values
- Remove Duplicate Records
- Validate Data Types
- Investigate Outliers
- Save Clean Dataset

## Import Required Libraries

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/creditcard.csv")
```

```python
rows, columns = df.shape

print(f"Rows    : {rows}")
print(f"Columns : {columns}")
```

```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_values
```

### Observation

No missing values were found in the dataset.

Therefore, no imputation or deletion is required.

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Transactions : {duplicates}")
```

```python
duplicate_transactions = df[df.duplicated()]

duplicate_transactions.head()
```

```python
df = df.drop_duplicates()
```

```python
print("Remaining duplicates:", df.duplicated().sum())
```

```python
rows, columns = df.shape

print(f"Rows after cleaning    : {rows}")
print(f"Columns after cleaning : {columns}")
```

```python
datatype_df = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values
})

datatype_df
```

```python
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
]

print(f"Potential Outliers : {len(outliers)}")
```

### Observation

The Amount feature contains several potential outliers.

These values are **not removed** because unusually large transactions may represent legitimate fraud cases.

Removing them could reduce the model's ability to detect fraudulent activity.

## Save Clean Dataset

```python
df.to_csv(
    "../data/processed/cleaned_creditcard.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
