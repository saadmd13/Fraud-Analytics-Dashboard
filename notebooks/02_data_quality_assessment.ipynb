# Data Quality Assessment

## Objective

Evaluate the quality of the dataset before performing data cleaning and exploratory data analysis.

### Quality Checks

- Missing Values
- Duplicate Records
- Class Imbalance
- Outlier Detection
- Data Types
- Feature Consistency
- Unique Values
- Memory Usage

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/creditcard.csv")
```

## Missing Value Analysis

```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing_values
```

Observation

The dataset contains no missing values.

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
duplicate_rows = df[df.duplicated()]

duplicate_rows.head()
```

```python
class_counts = df["Class"].value_counts()

class_counts
```

```python
class_percentage = (
    df["Class"].value_counts(normalize=True)*100
).round(4)

class_percentage
```

```python
unique_values = pd.DataFrame({
    "Unique Values": df.nunique()
})

unique_values
```

```python
memory = (
    df.memory_usage(deep=True) / 1024**2
).round(2)

memory
```

```python
print(
    f"Total Memory Usage : {memory.sum():.2f} MB"
)
```

```python
datatype_df = pd.DataFrame({
    "Column": df.columns,
    "Datatype": df.dtypes.values
})

datatype_df
```

```python
Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[
    (df["Amount"] < lower) |
    (df["Amount"] > upper)
]

print("Outliers :", len(outliers))
```

# Data Quality Summary

## Findings

- No missing values.
- Duplicate transactions exist.
- Dataset is highly imbalanced.
- Amount column contains outliers.
- Data types are appropriate.
- Memory usage is acceptable.
