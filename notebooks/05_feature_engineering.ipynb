# Feature Engineering

## Objective

Prepare the cleaned credit card transaction dataset for machine learning.

### Tasks

- Load cleaned dataset
- Create useful features
- Split features and target
- Scale numerical data
- Handle class imbalance
- Save preprocessing artifacts

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import joblib
```

```python
df = pd.read_csv("../data/processed/cleaned_creditcard.csv")
```

```python
print(df.shape)

df.head()
```

```python
X = df.drop("Class", axis=1)

y = df["Class"]
```

The machine learning model learns from the features (X) and predicts the target (y).

X → Transaction information

y → Fraud (0 or 1)

```python
summary = pd.DataFrame({
    "Data Type": X.dtypes,
    "Missing Values": X.isnull().sum(),
    "Unique Values": X.nunique()
})

summary
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## Observation

The dataset has been split using stratified sampling.

This preserves the fraud-to-genuine ratio in both the training and testing datasets, which is critical for imbalanced classification problems.

```python
print("Training Set :", X_train.shape)

print("Testing Set :", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

We fit the scaler only on the training dataset to avoid data leakage.

The testing dataset is transformed using the same scaler.

```python
feature_names = X.columns.tolist()

print(feature_names)
```

```python
print(X_train_scaled[:5])
```

```python
print(y_train.value_counts())
```

The dataset is highly imbalanced.

However, balancing techniques such as SMOTE will only be applied during model training to avoid information leakage.

```python
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)
```
